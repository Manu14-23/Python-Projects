```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf

tickers = ["SPY" , "EWU" , "URTH" , "TLT" , "GLD" , "QQQ"]
start_date = "2021-08-25"
end_date = "2026-08-25"

raw_data = yf.download(tickers = tickers , start = start_date, end = end_date)
adj_close = raw_data['Close']
adj_close = adj_close.dropna()
returns = adj_close.pct_change().dropna()

annual_returns = returns.mean() * 252
annual_cov_matrix = returns.cov() * 252
```

```python
risk_free_rate = 0.04

# Function to output the -ve Sharpe Ratio
def get_negative_sharpe(weights):
    port_return = weights.T @ annual_returns
    port_variance = weights.T @ annual_cov_matrix @ weights
    port_volatility = np.sqrt(port_variance)

    sharpe_ratio = (port_return - risk_free_rate) / port_volatility

    return -1 * sharpe_ratio


bounds = tuple((0, 1) for _ in range (len(returns.columns)))
constraints = { 'type': 'eq' , 'fun': lambda w:np.sum(w) -1}
```
